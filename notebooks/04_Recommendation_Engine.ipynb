```python
# Milestone 3 - Rule-Based Recommendation Engine

def generate_recommendations(
    nitrogen_gap,
    phosphorus_gap,
    potassium_gap,
    pH=None,
    organic_matter=None
):

    recommendations = []
    management_actions = []

    # -----------------------------
    # Nitrogen
    # -----------------------------
    if nitrogen_gap > 0:
        recommendations.extend(
            agricultural_knowledge_base["Nitrogen"]["recommendation"]
        )
        management_actions.extend(
            agricultural_knowledge_base["Nitrogen"]["management"]
        )

    # -----------------------------
    # Phosphorus
    # -----------------------------
    if phosphorus_gap > 0:
        recommendations.extend(
            agricultural_knowledge_base["Phosphorus"]["recommendation"]
        )
        management_actions.extend(
            agricultural_knowledge_base["Phosphorus"]["management"]
        )

    # -----------------------------
    # Potassium
    # -----------------------------
    if potassium_gap > 0:
        recommendations.extend(
            agricultural_knowledge_base["Potassium"]["recommendation"]
        )
        management_actions.extend(
            agricultural_knowledge_base["Potassium"]["management"]
        )

    # -----------------------------
    # pH
    # -----------------------------
    if pH is not None:

        if pH < 6.0:
            recommendations.extend(
                agricultural_knowledge_base["Acidic_pH"]["recommendation"]
            )
            management_actions.extend(
                agricultural_knowledge_base["Acidic_pH"]["management"]
            )

        elif pH > 8.0:
            recommendations.extend(
                agricultural_knowledge_base["Alkaline_pH"]["recommendation"]
            )
            management_actions.extend(
                agricultural_knowledge_base["Alkaline_pH"]["management"]
            )

    # -----------------------------
    # Organic Matter
    # -----------------------------
    if organic_matter is not None and organic_matter < 0.5:
        recommendations.extend(
            agricultural_knowledge_base["Low_Organic_Matter"]["recommendation"]
        )
        management_actions.extend(
            agricultural_knowledge_base["Low_Organic_Matter"]["management"]
        )

    # -----------------------------
    # Balanced nutrients
    # -----------------------------
    if (
        nitrogen_gap <= 0
        and phosphorus_gap <= 0
        and potassium_gap <= 0
    ):
        recommendations.extend(
            agricultural_knowledge_base["Balanced_Nutrients"]["recommendation"]
        )
        management_actions.extend(
            agricultural_knowledge_base["Balanced_Nutrients"]["management"]
        )

    # Remove duplicate recommendations
    recommendations = list(dict.fromkeys(recommendations))
    management_actions = list(dict.fromkeys(management_actions))

    return {
        "Recommendations": recommendations,
        "Management Actions": management_actions
    }

print("Rule-based recommendation engine created successfully.")
```

```python
# Milestone 3 - Crop Suitability Scoring Engine

def calculate_crop_suitability(soil_type, pH):

    crop_results = []

    for crop, requirements in crop_knowledge_base.items():

        # -----------------------------
        # Soil type score
        # -----------------------------
        if soil_type in requirements["soil_types"]:
            soil_type_score = 60
        else:
            soil_type_score = 0

        # -----------------------------
        # pH score
        # -----------------------------
        if pH is not None:
            if (
                requirements["pH_min"]
                <= pH
                <= requirements["pH_max"]
            ):
                ph_score = 40
            else:
                ph_score = 0
        else:
            ph_score = 0

        # -----------------------------
        # Total suitability score
        # -----------------------------
        total_score = soil_type_score + ph_score

        # -----------------------------
        # Suitability category
        # -----------------------------
        if total_score >= 80:
            category = "Highly Suitable"
        elif total_score >= 60:
            category = "Moderately Suitable"
        else:
            category = "Less Suitable"

        crop_results.append({
            "Crop": crop,
            "Soil Type Score": soil_type_score,
            "pH Score": ph_score,
            "Suitability Score": total_score,
            "Suitability": category
        })

    # Sort highest score first
    crop_results = sorted(
        crop_results,
        key=lambda x: x["Suitability Score"],
        reverse=True
    )

    return pd.DataFrame(crop_results)


print("Crop suitability scoring engine created successfully.")
```

```python
# Relative nutrient requirement framework for crop suitability
# H = Higher importance, M = Moderate importance

crop_nutrient_requirements = {
    "Rice": {
        "N": "H",
        "P": "M",
        "K": "H"
    },
    "Maize": {
        "N": "H",
        "P": "M",
        "K": "M"
    },
    "Wheat": {
        "N": "H",
        "P": "M",
        "K": "M"
    },
    "Groundnut": {
        "N": "M",
        "P": "H",
        "K": "M"
    },
    "Cotton": {
        "N": "H",
        "P": "M",
        "K": "H"
    },
    "Chickpea": {
        "N": "M",
        "P": "H",
        "K": "M"
    },
    "Sugarcane": {
        "N": "H",
        "P": "M",
        "K": "H"
    }
}

print("Crop nutrient requirement framework created successfully.")
print("\nCrops included:", list(crop_nutrient_requirements.keys()))
```

```python
def calculate_nutrient_suitability(
    nitrogen_gap,
    phosphorus_gap,
    potassium_gap,
    crop
):
    gaps = {
        "N": nitrogen_gap,
        "P": phosphorus_gap,
        "K": potassium_gap
    }

    importance_weights = {
        "H": 1.0,
        "M": 0.5
    }

    requirements = crop_nutrient_requirements[crop]

    total_weight = sum(
        importance_weights[requirements[nutrient]]
        for nutrient in ["N", "P", "K"]
    )

    achieved_weight = 0

    for nutrient in ["N", "P", "K"]:
        if gaps[nutrient] <= 0:
            # Nutrient is adequate
            achieved_weight += importance_weights[
                requirements[nutrient]
            ]
        else:
            # Nutrient has a positive gap
            # Give partial suitability credit
            achieved_weight += 0.5 * importance_weights[
                requirements[nutrient]
            ]

    nutrient_score = (achieved_weight / total_weight) * 30

    return round(nutrient_score, 2)


print("Nutrient suitability scoring function created successfully.")
```

```python
def calculate_final_crop_suitability(
    soil_type,
    pH,
    nitrogen_gap,
    phosphorus_gap,
    potassium_gap
):
    crop_results = []

    for crop, requirements in crop_knowledge_base.items():

        # 1. Soil type score: 40 points
        if soil_type in requirements["soil_types"]:
            soil_type_score = 40
        else:
            soil_type_score = 0

        # 2. pH score: 30 points
        if (
            pH is not None
            and requirements["pH_min"] <= pH <= requirements["pH_max"]
        ):
            ph_score = 30
        else:
            ph_score = 0

        # 3. Nutrient score: 30 points
        nutrient_score = calculate_nutrient_suitability(
            nitrogen_gap,
            phosphorus_gap,
            potassium_gap,
            crop
        )

        # Final score
        total_score = (
            soil_type_score +
            ph_score +
            nutrient_score
        )

        # Suitability category
        if total_score >= 80:
            category = "Highly Suitable"
        elif total_score >= 60:
            category = "Moderately Suitable"
        elif total_score >= 40:
            category = "Less Suitable"
        else:
            category = "Not Suitable"

        crop_results.append({
            "Crop": crop,
            "Soil Type Score": soil_type_score,
            "pH Score": ph_score,
            "Nutrient Score": nutrient_score,
            "Suitability Score": round(total_score, 2),
            "Suitability": category
        })

    # Rank crops from highest to lowest
    results = pd.DataFrame(crop_results)

    results = results.sort_values(
        by="Suitability Score",
        ascending=False
    ).reset_index(drop=True)

    return results


print("Final crop suitability scoring engine created successfully.")
```

```python
def calculate_soil_health_score(
    nitrogen_gap,
    phosphorus_gap,
    potassium_gap,
    pH=None,
    organic_matter=None
):
    # ------------------------------------------------
    # 1. NPK score: 60 points
    # ------------------------------------------------
    nutrient_gaps = [
        nitrogen_gap,
        phosphorus_gap,
        potassium_gap
    ]

    positive_gaps = sum(gap > 0 for gap in nutrient_gaps)

    if positive_gaps == 0:
        nutrient_score = 60
    elif positive_gaps == 1:
        nutrient_score = 40
    elif positive_gaps == 2:
        nutrient_score = 20
    else:
        nutrient_score = 0

    # ------------------------------------------------
    # 2. pH score: 20 points
    # ------------------------------------------------
    if pH is None:
        ph_score = 20
    elif 6.0 <= pH <= 7.5:
        ph_score = 20
    elif 5.5 <= pH < 6.0 or 7.5 < pH <= 8.0:
        ph_score = 10
    else:
        ph_score = 0

    # ------------------------------------------------
    # 3. Organic matter score: 20 points
    # ------------------------------------------------
    if organic_matter is None:
        om_score = 20
    elif organic_matter >= 1.0:
        om_score = 20
    elif organic_matter >= 0.5:
        om_score = 10
    else:
        om_score = 0

    # ------------------------------------------------
    # Final score
    # ------------------------------------------------
    total_score = nutrient_score + ph_score + om_score

    # Health category
    if total_score >= 80:
        health_level = "Good"
    elif total_score >= 60:
        health_level = "Moderate"
    elif total_score >= 40:
        health_level = "Poor"
    else:
        health_level = "Very Poor"

    return {
        "Nutrient Score": nutrient_score,
        "pH Score": ph_score,
        "Organic Matter Score": om_score,
        "Soil Health Score": total_score,
        "Soil Health Level": health_level
    }


print("Soil health scoring function created successfully.")
```

```python
def generate_complete_soil_analysis(
    soil_type,
    nitrogen_gap,
    phosphorus_gap,
    potassium_gap,
    pH=None,
    organic_matter=None
):
    # ------------------------------------------------
    # 1. Soil Health Score
    # ------------------------------------------------
    health_result = calculate_soil_health_score(
        nitrogen_gap=nitrogen_gap,
        phosphorus_gap=phosphorus_gap,
        potassium_gap=potassium_gap,
        pH=pH,
        organic_matter=organic_matter
    )

    # ------------------------------------------------
    # 2. Recommendation Engine
    # ------------------------------------------------
    recommendation_result = generate_complete_recommendation(
        soil_type=soil_type,
        nitrogen_gap=nitrogen_gap,
        phosphorus_gap=phosphorus_gap,
        potassium_gap=potassium_gap,
        pH=pH,
        organic_matter=organic_matter,
        soil_health_score=health_result["Soil Health Score"],
        soil_health_level=health_result["Soil Health Level"]
    )

    # ------------------------------------------------
    # 3. Crop Suitability
    # ------------------------------------------------
    crop_results = calculate_final_crop_suitability(
        soil_type=soil_type,
        pH=pH,
        nitrogen_gap=nitrogen_gap,
        phosphorus_gap=phosphorus_gap,
        potassium_gap=potassium_gap
    )

    # ------------------------------------------------
    # 4. Final result
    # ------------------------------------------------
    return {
        "Soil Information": {
            "Soil Type": soil_type,
            "pH": pH,
            "Organic Matter": organic_matter
        },

        "Nutrient Gaps": {
            "Nitrogen": nitrogen_gap,
            "Phosphorus": phosphorus_gap,
            "Potassium": potassium_gap
        },

        "Nutrient Deficiencies":
            recommendation_result["Nutrient Deficiencies"],

        "Soil Health": health_result,

        "Recommendations":
            recommendation_result["Recommendations"],

        "Management Actions":
            recommendation_result["Management Actions"],

        "Crop Suitability":
            crop_results
    }


print("Complete Milestone 3 soil analysis engine created successfully.")
```
