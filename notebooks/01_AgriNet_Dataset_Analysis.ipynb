```python
numeric_cols = ["pH", "EC", "OC", "N", "P", "K", "S", "Zn", "B", "Fe", "Mn", "Cu"]

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: {count} possible outliers")
```

```python
numeric_cols = ["pH", "EC", "OC", "N", "P", "K",
                "S", "Zn", "B", "Fe", "Mn", "Cu"]

outlier_summary = {}

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    outlier_summary[col] = {
        "count": len(outliers),
        "lower_bound": lower,
        "upper_bound": upper
    }

outlier_summary
```

```python
outlier_table = pd.DataFrame(outlier_summary).T

outlier_table
```
